```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # or "yolov8s.pt" (smaller = faster on CPU)

# Train with CPU
results = model.train(
    data="data.yaml",
    epochs=50,  # Reduce epochs for faster CPU training
    imgsz=320,  # Smaller image size for CPU
    batch=4,    # Reduce batch size for CPU memory
    device="cpu",  # Force CPU usage
    name="hands_detection_cpu"
)
```

```python
# Add class weights to handle imbalance (if needed)
class_weights = [1.0, 1.2]  # Higher weight for "hands_off"

results = model.train(
    ...
    cls=class_weights,  # Class-wise loss weights
    fliplr=0.5,  # Horizontal flip probability
    mosaic=1.0,  # Mosaic augmentation
    mixup=0.2,   # Mixup augmentation
)
```

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Initialize model
model = YOLO("/Users/lakshmanmandapati/Downloads/YOLO(DD)/dataset/runs/detect/hands_detection_cpu/weights/best.pt")

# Your test images (replace with actual paths)
test_images = [
    "test_image_1.jpg",
    "test_image_2.jpg", 
    "test_image_3.jpg",
    "test_image_4.jpg",
    "test_image_5.jpg"
]

# Configure prediction
pred_args = {
    "conf": 0.5,
    "iou": 0.45,
    "show_labels": True,
    "show_conf": True
}

# Create display layout
fig = plt.figure(figsize=(18, 20))
gs = fig.add_gridspec(3, 4)

# Axes for individual images
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1]) 
ax3 = fig.add_subplot(gs[1, 0])
ax4 = fig.add_subplot(gs[1, 1])
ax5 = fig.add_subplot(gs[2, 0])

# Axes for aggregated results
ax_summary = fig.add_subplot(gs[:, 2:])

# Process and display each image
all_detections = []
axes = [ax1, ax2, ax3, ax4, ax5]

for idx, img_path in enumerate(test_images):
    results = model.predict(img_path, **pred_args)
    img_array = results[0].plot()[:, :, ::-1]  # Convert BGR to RGB
    
    # Display image
    axes[idx].imshow(img_array)
    axes[idx].axis('off')
    axes[idx].set_title(f"Image {idx+1}")
    
    # Store detections
    detections = results[0].boxes.data.cpu().numpy()
    if len(detections) > 0:
        df = pd.DataFrame(detections, columns=['x1','y1','x2','y2','conf','class'])
        all_detections.append(df)

# Combine all detections
if all_detections:
    combined_dets = pd.concat(all_detections)
    
    # Class distribution plot
    class_counts = combined_dets['class'].value_counts()
    colors = ['#4CAF50', '#FF5722']  # Green for hands-on, orange for hands-off
    class_counts.plot(kind='bar', ax=ax_summary, color=colors, width=0.6)
    
    # Summary table
    stats = {
        'Total Detections': len(combined_dets),
        'Hands On': len(combined_dets[combined_dets['class'] == 0]),
        'Hands Off': len(combined_dets[combined_dets['class'] == 1]),
        'Avg Confidence': f"{combined_dets['conf'].mean():.2f}"
    }
    
    ax_summary.table(cellText=[[stats[k]] for k in stats],
                   rowLabels=list(stats.keys()),
                   colWidths=[0.5],
                   cellLoc='center',
                   loc='bottom',
                   bbox=[0.1, -0.5, 0.8, 0.3])
    
    ax_summary.set_title("Aggregated Detection Results", pad=20)
    ax_summary.set_xticks([0, 1], ['Hands On', 'Hands Off'])
    ax_summary.set_ylabel("Count")
    ax_summary.grid(axis='y', linestyle='--', alpha=0.7)
else:
    ax_summary.text(0.5, 0.5, "No detections in any images", 
                  ha='center', va='center', fontsize=12)

plt.tight_layout()
plt.show()
```
